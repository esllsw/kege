```python
a = [list(map(float, x.split())) for x in open('27A_20971.txt')]

n = 4
cl = {c:[] for c in (1,2,3,4)}

for st in a:
    x,y = st
    if y < (2*x + 8):
        cl[1] += [st]
    else:
        cl[2] += [st]

def dist(a,b):
    x1, y1 = a
    x2, y2 = b
    return ((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5

def cent(cl):
    res = (0,0)
    sum_res = 10 ** 20
    for a in cl:
        sum_a = 0
        for b in cl:
            sum_a += dist(a,b)
        if sum_a < sum_res:
            sum_res = sum_a
            res = a
    return res

px, py = 0,0
for x in cl:
    center = cent(cl[x])
    px += center[0]
    py += center[1]

print(int((px / n) * 10000), int((py / n)* 10000))
```

```python
f = open('555.txt')
cl1,cl2,cl3,cl4=[],[],[],[]
for i in f:
    x,y=map(float,i.replace(',','.').split())
    if x>2 and y<2:
        cl1.append([x,y])
    elif x>2 and y>2:
        cl2.append([x,y])
    elif x<2 and y<0:
        cl3.append([x,y])
    elif x<2 and y>0:
        cl4.append([x,y])

def pl(cl):
    s=[]
    for i in range(len(cl)):
        k=0
        for j in range(len(cl)):
            x1,y1=cl[i]
            x2,y2=cl[j]
            if (((x2-x1)**2+(y2-y1)**2)**0.5)<1:
                k=k+1
        s.append(k)
    return sum(s)/len(s)
m1=pl(cl1)
m2=pl(cl2)
m3=pl(cl3)
m4=pl(cl4)
print(min(m1,m2,m3,m4)*100000,(m1+m2+m3+m4)/4*100000)
```
