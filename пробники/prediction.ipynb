```python
def tr(n):
    s = ''
    while n > 0:
        s = str(n % 3) + s
        n //= 3
    return s
res = []
for n in range(1,1000):
    s = tr(n)
    if n % 3 == 0: s += tr(n)[-2:]
    else: s += tr((n%3)*5)
    r = int(s,3)
    if r >= 290: res.append(n)
print(min(res))
```

```python
from itertools import product
s = '0123456'
cnt = 0
for i in product(s,repeat=6):
    a = ''.join(i)
    if a[0] != '0' and a.count('3') == 1:
        if (not '13' in a )and (not '31' in a):
            if (not '51' in a) and (not '15' in a):
                cnt += 1

print(cnt)
```

```python
f = open('9.csv')
cnt = 0
for line in f:
    a = [int(x) for x in line.split(',')]
    povt = [x for x in a if a.count(x) == 3]
    ne_povt = [x for x in a if a.count(x) == 1]
    if len(set(povt)) == 1 and len(set(ne_povt)) == 3:
        if povt[0] < min(ne_povt)*2:
            cnt += 1
print(cnt)
```

```python
from functools import lru_cache
@lru_cache(None)
def f(n):
    if n <= 1: return 1
    if n > 1: return n + f(n - 3)

for i in range(20000):
    f(i)
print(f(16025) - f(16021))
```

```python
f = open('17 (1).txt')
a = [int(x) for x in f]
ch = max([x for x in a if str(x)[-2:]== '23'])
cnt = mx = 0
for i in range(len(a) - 2):
    s = a[i:i+3]
    s = sorted(s)
    otr = [x for x in s if x < 0]
    poz = [x for x in s if x >= 0]
    if len(otr) == 1:
        if sum(poz) > ch:
            cnt += 1
            mx = max(mx, sum(poz))
print(cnt, mx)
```

```python
def f(s,m):
    if s < 88: return m % 2 == 0
    if m == 0: return 0
    nm = [f(s - 2, m - 1), f(s // 2, m - 1)]
    return any(nm) if (m - 1) % 2 == 0 else all(nm)

print(19, [s for s in range(89,200) if f(s,2)])
print(20, [s for s in range(89,200) if f(s,3) and not f(s,1)])
print(21, [s for s in range(89,200) if f(s,4) and not f(s,2)])
```

```python
def f(st,en):
    if st > en or st == 16: return 0
    if st == en: return 1
    return f(st + 1, en) + f(st + 3, en) + f(st * 2, en)
print(f(3,9)*f(9,30))
```
