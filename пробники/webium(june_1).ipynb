```python
def tr(x):
    s = ''
    while x > 0:
        s = str(x%3) + s
        x //= 3
    return s
res = []
for n in range(1,1000):
    s = tr(n)
    if n % 7 == 0: s += tr(n)[-2:]
    else: s += tr((n%7)*3)
    r = int(s,3)
    if r > 369: res.append(r)
print(min(res))
```

```python
def f(c, e, p):  # объявим функцию от трех переменных: с - начальная точка, е - конечная точка, р - предыдущая команда
    # т.к у нас есть команда, возвращающая нас назад, то конец увеличим, и если туда уходим, то возвращаем 0
    if c > e + 100: return 0
    if c == e: return 1  # если начальная точка равна конечной, то возвращаем единицу
    # если предыдущая команда была А, то возвращаем два возможных варианта: ходы В и С
    if p == 'a': return f(c * 2, e, 'b') + f(c * 3, e, 'c')
    # если предыдущая команда не А, то возвращаем все возможные ходы
    if p != 'a': return f(c - 1, e, 'a') + f(c * 2, e, 'b') + f(c * 3, e, 'c')

print(f(3, 15, ''))  # выводим ответ
```

```python
file_path = '26.txt'
# Открываем файл для чтения и обрабатываем его содержимое
with open(file_path, 'r') as file:
    n = int(file.readline().strip())  # Читаем первую строку, содержащую количество занятых мест
    occup = {}  # Словарь для хранения занятых позиций по рядам
    # Считываем файл построчно
    for line in file:
        # Превращаем каждую переменную в число, с помощью map(int, )
        row, pos = map(int, line.strip().split())
        if row in occup:
            occup[row].add(pos)
        else:
            occup[row] = {pos}

# Инициализация переменных для хранения результатов
maxrow = 0
maxcount = 0

# Обход всех рядов и их занятых позиций
for row, places in occup.items():
    # Сортируем позиции в каждом ряду
    sort_places = sorted(places)
    maxi = 0

    # Находим максимальное количество подряд идущих свободных мест в ряду
    for i in range(1, len(sort_places)):
        gap = sort_places[i] - sort_places[i - 1] - 1
        if gap > maxi:
            maxi = gap

    # Обновляем глобальные максимумы, если найденный локальный максимум больше или равен текущему максимуму
    # и номер ряда больше предыдущего найденного максимального номера ряда
    if maxi > maxcount or (maxi == maxcount and row > maxrow):
        maxcount = maxi
        maxrow = row

print(maxrow, maxcount)
```

```python
# Считываем звезды из файла
file = open('27b.txt')  # Путь к файлу. Не забываем поменять его на нужный путь к файлу B при замене блока с кластеризацией
xy = file.readline()  # Пропуск строки с символами x и y
stars = [list(map(float, x.replace(',', '.').split())) for x in file]  # Считываем данные о звездах как списки с координатами x и y в список stars

# Кластеризация на основе текстовых файлов (Для файла A)
clusters = {1: [], 2: [], 3:[], 4:[]}  # Задаем словарь с количеством кластеров из файла
for star in stars:  # Проходим по всем звездам в файле
    x, y = star  # Отделяем координаты x и y
    if x < 2:
        if y > 0: clusters[1] += [star]
        else: clusters[2] += [star]
    else:
        if y > 2: clusters[3] += [star]
        else: clusters[4] += [star]

# Кластеризация на основе текстовых файлов (Для файла B). Комментируем кластеризацию
# для файла A и раскомментируем код ниже
# clusters = {1: [], 2: [], 3:[], 4:[], 5:[], 6:[], 7:[]}  # Задаем словарь с количеством кластеров из файла
# for star in stars:  # Проходим по всем звездам в файле
#     x, y = star  # Отделяем координаты x и y
#     if x < -4: clusters[1] += [star]
#     elif x < 1:
#         if y > -3: clusters[2] += [star]
#         else: clusters[3] += [star]
#     elif x < 6:
#         if y > 1: clusters[4] += [star]
#         else: clusters[5] += [star]
#     else:
#         if y > 0: clusters[6] += [star]
#         else: clusters[7] += [star]

# Считаем расстояние по функции из условия
def r(a, b):  # Задаем функцию для точек A и B
    x1, y1 = a  # Точка A с координатами x1 и y1
    x2, y2 = b  # Точка B с координатами x2 и y2
    return ((x2 - x1)**2 + (y2 - y1)**2) ** 0.5  # Возвращаем расстояние, посчитанное по условию

# Считаем среднее расстояние кластеров
def sr_r(cl):
    s = 0  # Создадим переменную для суммы расстояний между точками
    k = 0  # И переменную для количества пар точек
    for p in cl:  # Проходимся по кластеру
        for p1 in cl:  # Перебираем точки в кластере
            if p != p1:  # Проверяем, что мы рассматриваем разные точки
                s += r(p, p1)  # Добавляем в общую сумму текущее расстояние
                k += 1  # И обновляем количество пар
    return s/k  # Возвращаем среднее расстояние кластера

res = []  # Создадим список для средних расстояний всех кластеров
for cl in clusters:  # Проходим по кластерам
    d = sr_r(clusters[cl])  # Считаем среднее расстояние каждого кластера
    res.append(d)  # Добавляем текущее расстояние в список

print(int(min(res)*100_000), int(max(res)*100_000))  # Выводим ответ
```
