```python
#8
from itertools import product

gr = ['35','36','37','38','53','63','73','83']
s = '012345678'
cnt = 0
for i in product(s,repeat=5):
    a = ''.join(i)
    c = 0
    for q in gr:
        if q in a: c += 1
    if a[0] != '0' and a.count('3') == 1:
        if c == 0: cnt += 1
print(cnt)
```

```python
#19-21
def f(s1,s2,m):
    if (s1 + s2) >= 231: return m % 2 == 0
    if m == 0: return 0
    nm = [f(s1 + 1, s2, m - 1), f(s1 * 2, s2, m - 1), f(s1, s2 + 1, m - 1), f(s1, s2 * 2, m - 1)]
    return any(nm) if (m - 1) % 2 == 0 else all(nm)

#print(19, [s for s in range(1,214) if f(17, s, 2)])
print(20, [s for s in range(1,214) if f(17, s, 3) and not f(17,s, 1)])
print(21, [s for s in range(1,214) if f(17, s, 4) and not f(17,s, 2)])
```

```python
#23
def f(st, en):
    if st == en: return 1
    if en > st: return 0
    return f(st - 1, en) + f(st // 2, en)

print(f(30,13) * f(13,1))
```

```python
#25
def divs(n):
    res = set()
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            res.add(i)
            res.add(n // i)
    return sorted(res)

def prime(n):
    if n < 2: return False
    for i in range(2,int(n**0.5) + 1):
        if n % i == 0: 
            return False
    return True

for n in range(10 ** 9, 10 ** 10 + 1):
    a = divs(n)
    if len(a) >= 3:
        sm = (a[0] + a[-2] + a[-1])
        if prime(sm):
            print(n,sm)
```

```python
#27
f = open('27a.txt')

k1,k2 = [], []
for line in f:
    x,y = map(float, line.split())
    if x < 8:
        k1.append([x,y])
    else:
        k2.append([x,y])

def dist1(a,b):
    return ((b[0] - a[0]) ** 2 + (b[1] - a[1]) ** 2) ** 0.5

def dist2(a,b):
    return (abs(b[0] - a[0]) + abs(b[1] - a[1]))

def centroid1(cluster):
    sum_res = 10 ** 20
    final = []
    for a in cluster:
        sum_dist = sum(dist1(point, a) for point in cluster)
        if sum_dist < sum_res:
            sum_res = sum_dist
            final = a
    return final

def centroid2(cluster):
    sum_res = 10 ** 20
    final = []
    for a in cluster:
        sum_dist = sum(dist2(point, a) for point in cluster)
        if sum_dist < sum_res:
            sum_res = sum_dist
            final = a
    return final

center1 = [centroid1(cluster) for cluster in (k1,k2)]
center2 = [centroid2(cluster) for cluster in (k1,k2)]
print(center1)
print(center2)
```
