```python
def sv(n):
    s = ''
    while n > 0:
        s = str(n % 7) + s
        n //= 7
    return s

for x in range(2301,1,-1):
    s = 7 ** 350 + 7 ** 150 - x
    a = sv(s)
    if a.count('0') == 200:
        print(x)
        break
```

```python
def f(s,m):
    if s >= 67: return m % 2 == 0
    if m == 0: return 0
    nm = [f(s + 1,m - 1), f(s + 4, m - 1),f(s * 3,m - 1)]
    return any(nm) if (m - 1) % 2 == 0 else all(nm)

print(19, [s for s in range(1,67) if f(s,2)])
print(20, [s for s in range(1,67) if f(s,3) and not f(s,1)])
print(21, [s for s in range(1,67) if f(s,4) and not f(s,2)])
```

```python
def f(st,en):
    if st == en: return 1
    if st > en or st == 8: return 0
    return f(st + 1, en) + f(st + 2, en) + f(st * 2, en)
print(f(3,14)*f(14,18))
```

```python
def divs(x):
    res = set()
    for i in range(1,int(x**0.5)+1):
        if x % i == 0:
            res.add(i)
            res.add(x//i)
    return sorted(res)

for i in range(500_000, 1_000_000):
    if sum(divs(i)) % 10 == 6:
        print(i, sum(divs(i)))
```

```python
f = open('27_A_21911.txt')

k1, k2 = [], []
for line in f:
    x, y = map(float,line.split())
    if 4 < y < 12:
        k1.append([x,y])
    else:
        k2.append([x,y])

# f = open('27_B_21911.txt')

# k1, k2, k3 = [], [], []
# for line in f:
#     x, y = map(float,line.split())
#     if x < 10:
#         k1.append([x,y])
#     if x > 20:
#         k2.append([x,y])
#     else:
#         k3.append([x,y])

def dist(a,b):
    return ((b[0] - a[0]) ** 2 + (b[1] - a[1]) ** 2) ** 0.5

def centroid(cluster):
    res_sum = 10 ** 20
    final = []
    for a in cluster:
        sum_dist = sum(dist(point,a) for point in cluster)
        if sum_dist < res_sum:
            res_sum = sum_dist
            final = a
    return final

center = [centroid(cluster) for cluster in (k1,k2)]

px = sum(x for x, y in center) / len(center) * 10000
py = sum(y for x, y in center) / len(center) * 10000

print(px,py)
```

```python
f = open('17_21903.txt')
a = [int(x) for x in f]
m = min([x for x in a if abs(x) % 100 == 15 and len(str(abs(x))) == 3]) ** 2

cnt = 0
mn = 10 ** 10

for i in range(len(a) - 2):
    s = a[i:i+3]
    s.sort()
    if (s[0] > 0 and s[1] > 0 and s[2] > 0) or (s[0] < 0 and s[1] < 0 and s[2] < 0):
        if s[0] * s[-1] > m:
            cnt += 1
            mn = min(mn, s[0] * s[-1])

print(cnt, mn)
```
