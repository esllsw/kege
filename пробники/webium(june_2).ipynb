```python
import turtle as t
t.speed(10)
t.tracer(0)
t.screensize(2000,2000)
k = 25

for _ in range(4):
    t.forward(4 * k)
    t.left(270)

t.up()
for _ in range(3):
    t.forward(k)
    t.left(270)
    t.forward(k)
    t.left(90)

t.down()
for _ in range(2):
    t.forward(9 * k)
    t.left(270)
    t.forward(11 * k)
    t.left(270)

t.up()
for x in range(-1, 19):
    for y in range(-20, 2):
        t.goto(x * k, y * k)
        t.dot(5, 'red')
t.done()
```

```python
from functools import lru_cache
@lru_cache(None)
def f(n):
    if n == 0: return 0
    elif n % 2 == 0: return f(n-1) + 1
    else: return f(n //2)


cnt = 0
for i in range(10**9):
    if f(i) == 2:
        cnt+= 1
        print(i, cnt)
```

```python
s = open('24.txt').read()
cntX = cntY = l = k = r = 0

while r < len(s):
    if s[r] == 'X': cntX += 1
    if s[r] == 'Y': cntY += 1
    while cntY > 1 or cntX > 1:
        if s[l] == 'X':
            cntX -= 1    
        if s[l] == 'Y':
            cntY -= 1
        l += 1
    k = max(k,r - l + 1)
    r += 1
print(k)
```

```python
def divs(n):
    res = set()
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            res.add(i)
            res.add(n//i)
    return sorted(res,reverse=True)

for x in range(300_000_000,400_000_000):
    if len(divs(x)) >= 6:
        print(x, divs(x)[5])
```

```python
f = open('27B.txt')

k1,k2,k3,k4,k5 = [], [], [], [], []
for line in f:
    x, y = map(float,line.split())
    if x < -3.5 and y < 2: k1.append([x,y])
    elif x < -1.5 and y > 2: k2.append([x,y])
    elif -3 < x < 0 and y < 2: k3.append([x,y])
    elif x > 0 and y < 2: k4.append([x,y])
    else: k5.append([x,y])
```

```python
def dist(a,b):
    return ((b[0] - a[0]) ** 2 + (b[1] - a[1])**2) ** 0.5

def centroid(cluster):
    sum_res = -10**10
    final = []
    for a in cluster:
        sum_dist = sum(dist(point,a) for point in cluster)
        if sum_dist > sum_res:
            sum_res = sum_dist
            final = a
    return final

center = [centroid(cluster) for cluster in (k1, k2, k3, k4, k5)]

px = sum(x for x, y in center) / len(center) * 10000
py = sum(y for x, y in center) / len(center) * 10000

print(int(px), int(py))
```
