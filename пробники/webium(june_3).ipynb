```python
import turtle as t
t.tracer(0)
t.screensize(2000,2000)
k = 25

for _ in range(4):
    t.fd(10 * k)
    t.rt(270)

t.up()
t.fd(3 * k)
t.rt(270)
t.fd(5 * k)
t.rt(90)

t.down()
for _ in range(2):
    t.fd(10 * k)
    t.rt(270)
    t.fd(12 * k)
    t.rt(270)

t.up()
for x in range(-5,15):
    for y in range(0, 20):
        t.goto(x * k, y * k)
        t.dot(5, 'red')
t.done()
```

```python
from itertools import product
s = '012345678'
cnt = 0
for i in product(s,repeat=4):
    a = ''.join(i)
    if a[0] != '0':
        if a.count('8') == 1: 
            if a.index('8') == 2:
                if (int(a[0]) + int(a[1])) == int(a[3]): cnt += 1
            elif a.index('8') == 1:
                if int(a[0]) == (int(a[2]) + int(a[3])): cnt += 1
            elif a.index('8') == 0: 
                if (int(a[3]) + int(a[1]) + int(a[2])) == 0: cnt += 1
print(cnt)
            
```

```python
res = []
for n in range(4,10000):
    s = '3' + '5' * n
    while '333' in s or '555' in s:
        if '555' in s: s = s.replace('555', '3', 1)
        elif '333' in s: s = s.replace('333', '5', 1)
    q = sum(map(int,s))
    res.append(q)
print(max(res))
```

```python
f = open('17.txt')
a = [int(x) for x in f]
r = 0
cnt = 0
mn = 10 ** 10
for i in a:
    r += sum(map(int,str(i)))
for q in range(len(a) - 1):
    s = a[q:q+2]
    if (s[0] % 100 == 10 and s[1] % 100 != 10) or (s[1] % 100 == 10 and s[0] % 100 != 10):
        if sum(s) < r:
            cnt += 1
            mn = min(mn,sum(s))
print(cnt, mn)
```

```python
def f(s, m):
    if s >= 469: return m % 2 == 0
    if m == 0: return 0
    nm = [f(s + 5, m - 1), f(s * 3, m - 1)]
    return any(nm) if (m-1) % 2 == 0 else all(nm)

print(19, [s for s in range(1,469) if f(s,2)])
print(20, [s for s in range(1,469) if f(s,3) and not f(s,1)])
print(21, [s for s in range(1,469) if f(s,4) and not f(s,2)])
```

```python
from functools import lru_cache

# Теперь прописываем функцию из условия
@lru_cache(None)
def f(n):
    if n >= 2025:
        return n
    return f(n+1) - f(n+2) + 7

# Если мы попробуем запустить программу в таком виде, то ничего не получится
# Сперва нужно запомнить значения f(n), а уже потом вызывать функцию
for i in range(2024, 14, -1):
    f(i)

# А теперь выводим ответ
print(f(15)-f(24))
```
