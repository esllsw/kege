```python
def tw(n):
    s = ''
    while n > 0:
        if n%12 == 11: q = 'B'
        elif n % 12 == 10: q = 'A'
        else: q = str(n % 12)
        s = q + s
        n //= 12
    return s

w = mx1 = mx2 = 0
for x in range(1,3500):
    s = tw(12 ** 457 + 12 ** 48 - x)
    mx1 = s.count('0')
    if mx1 > mx2:
        mx2 = mx1
        w = x
    print(w,mx2)
```

```python
def divs(x):
    res = set()
    for i in range(2,int(x**0.5)+1):
        if x % i == 0:
            res.add(i)
            res.add(x//i)
    return sorted(res)

for x in range(625681,758642):
    if len(divs(x)) == 7 and divs(x)[0] >= 10:
        print(x, divs(x))
```

```python
# f = open('27_А.txt')
# k1, k2 = [], []
# for line in f:
#     x, y = map(float,line.split())
#     if x > 3.6: k1.append([x,y])
#     else: k2.append([x,y])

f = open('27_Б.txt')
k1, k2, k3 = [], [], []
for line in f:
    x,y = map(float,line.split())
    if x < -2 * y + 4.5: k1.append([x,y])
    elif x < 4 * y + 6: k2.append([x,y])
    else: k3.append([x,y])

def dist(a,b):
    return ((b[0] - a[0]) ** 2 + (b[1] - a[1]) ** 2) ** 0.5

def centroid(cluster):
    sum_res = 10**10
    final = []
    for a in cluster:
        sum_dist = sum(dist(point,a) for point in cluster)
        if sum_dist < sum_res:
            sum_res = sum_dist
            final = a
    return final

center = [centroid(cluster) for cluster in (k1,k2, k3)]
px = sum(x for x, y in center) / len(center) * 10000
py= sum(y for x, y in center) / len(center) * 10000

print(int(px), int(py))
```

```python
from math import dist


with open("27_Б.txt") as file:
    header = file.readline().split()
    data = [list(map(float, line.split())) for line in file]


clusters = []
while len(data):
    cl = [data.pop()]
    for p in cl:
        for p1 in data:
            if dist(p, p1) < 0.3:
                cl.append(p1)
                data.remove(p1)
    clusters.append(cl)

print([len(c) for c in clusters])

xmins, ymins = 0, 0
for cl in clusters:
    min_dist = float("inf")
    for p in cl:
        dist_ = 0
        for p1 in cl:
            dist_ += dist(p, p1)

        if dist_ < min_dist:
            min_dist = dist_
            xmin, ymin = p

    xmins += xmin
    ymins += ymin

print(int(xmins / len(clusters) * 10000), int(ymins / len(clusters) * 10000))
```
