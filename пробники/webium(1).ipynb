```python
#5
def sev(n):
    s = ''
    while n > 0:
        s = str(n % 7) + s
        n //= 7
    return s
res = []
for n in range(343,2402):
    s = sev(n)
    if int(s) % 2 == 0: s = '6' + s
    else: s = '5' + s
    r = int(s,7)
    if r > 14500:
        res.append(n)
print(len(res))
```

```python
#6
import turtle as t
t.speed(10)
t.tracer(-1)
k = 15

for _ in range(2):
    t.forward(10 * k)
    t.right(90)
    t.forward(18 * k)
    t.right(90)

t.up()
t.forward(5 * k)
t.right(90)
t.fd(7 * k)
t.left(90)

t.down()
for _ in range(2):
    t.forward(10 * k)
    t.right(90)
    t.forward(7 * k)
    t.right(90) 

t.up()
for x in range(0,20):
    for y in range(-20,10):
        t.goto(x * k, y * k)
        t.dot(4, 'red')

t.done()
```

```python
#8
from itertools import product
comb1 = set()
for x in '0246':
    for y in '0246':
        for z in '0246':
            comb1.add(x+y+z)
comb2 = set()
for x in '0246':
    for y in '0246':
        comb2.add(x+y)
s = '0123456'
res = set()
for x in product(s,repeat=5):
    a = ''.join(x)
    b = a
    for i in comb1:
        if i in b:
            b = b.replace(i,'*')
    for p in comb2:
        if p in b: b = b.replace(p,'+')
    if b.count('*') == 0 and b.count('+') == 2:
        res.add(a)
print(len(res))
    
```

```python
from functools import lru_cache
@lru_cache(None)
def f(n):
    if n < 3: return 1
    if n > 2 and n % 2 == 1: return f(n - 1) + n
    else: return f(n-3) + 2 * n

for i in range(2100):
    f(i)
print(f(2048) - f(2041))
```

```python
f = open('17.txt')
a = [int(i) for i in f]
d = sum([i for i in a if abs(i) % 1000 == 111])
res = 0
sm = 10 ** 10
for i in range(len(a) - 2):
    s = a[i:i+3]
    s.sort()
    if s[0] % s[-1] == 0 or s[1] % s[-1] == 0:
        if s[0] * s[1] * s[2] >= d:
            res += 1
            sm = min(sm,s[0] * s[1] * s[2])
print(res,sm)
```

```python
def f(s,m):
    if s >= 199: return m % 2 == 0
    if m == 0: return 0
    nm = [f(s + 1, m - 1), f(s + 5, m - 1), f(s * 3, m - 1)]
    return any(nm) if (m-1) % 2 == 0 else all(nm)
print(19, [s for s in range(1, 199) if f(s, 2)])
print(20, [s for s in range(1, 199) if f(s, 3) and not f(s, 1)])
print(21, [s for s in range(1, 199) if (not f(s, 2)) and f(s, 4)])
```

```python
def f(st,en,cnt_p=0,cnt_u=0):
    if st == en:
        if cnt_p < cnt_u:
            return 1
    if st > en: return 0
    return f(st + 1, en, cnt_p + 1,cnt_u) + f(st * 2,en, cnt_p, cnt_u + 1) + f(st * 5,en,cnt_p,cnt_u + 1)
print(f(3,260))
```

```python
s = open('24.txt').readline()

# Убираем из строки все участки, где подряд идут два знака операций — это недопустимые выражения
# Заменяем такие участки пробелами, чтобы потом по ним удобно было разделять на фрагменты
s = s.replace('++', ' ').replace('+*', ' ').replace('*+', ' ').replace('**', ' ')

# Переменная для хранения максимальной длины подходящего выражения
m = 0

# Разбиваем строку по пробелам — получаем только те участки, где нет подряд идущих операторов
for c in s.split():
    # Если фрагмент достаточно длинный, проверяем его более детально
    if len(c) > m:
        # Перебираем все возможные позиции начала подстроки внутри текущего фрагмента
        for i in range(len(c)):
            # Начинаем только с цифр — если с символа операции, пропускаем
            if c[i] not in '*+':
                sub = ''
                # Постепенно строим подстроку, добавляя символы справа
                for j in range(i, len(c)):
                    sub += c[j]
                    try:
                        # Проверяем: если текущий символ — цифра (а не операция),
                        # и значение выражения sub равно 0 — обновляем максимум
                        if c[j] not in '*+' and eval(sub) == 0:
                            m = max(m, len(sub))
                            print(m,sub)
                    except:
                        # Если выражение некорректное (например, начинается с * или +, или содержит
                        # невалидную конструкцию типа "02") — просто пропускаем
                        pass
print(m)
```
