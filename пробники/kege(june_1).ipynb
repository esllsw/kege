```python
res = []
for n in range(1,1000):
    s = bin(n)[2:]
    if s.count('1') % 2 == 0:
        s += '1'
        s = '11' + s[2:]
    else:
        if s.count('0') < s.count('1'): s += '0'
        else: s += '1'
    r = int(s,2)
    if r > 271: res.append(n)

print(min(res))
```

```python
from math import dist
with open('27A_22623.txt') as file:
    data = [list(map(float,line.split())) for line in file]

clusters = []
while len(data):
    cl = [data.pop()]
    for p in cl:
        for p1 in data:
            if dist(p, p1) < 4:
                cl.append(p1)
                data.remove(p1)
    clusters.append(cl)

print([len(c) for c in clusters])

def centroid(cluster):
    sum_res = 10 ** 10
    final = []
    for a in cluster:
        sum_dist = sum(dist(point, a) for point in cluster)
        if sum_dist < sum_res:
            sum_res = sum_dist
            final = a
    return final

center = [centroid(cluster) for cluster in clusters]
p1 = 203 / 16
p2 = 276 / 16
p3 = 199 / 16
p = (p1 + p2 + p3) / 3
s = dist(center[1],center[2])
print(int(p * 1000), int(s * 1000))
```
